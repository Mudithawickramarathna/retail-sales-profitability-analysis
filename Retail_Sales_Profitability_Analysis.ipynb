```python
import pandas as pd

url = "https://raw.githubusercontent.com/Victory-Data/Superstore-Sales-Dataset/main/Sample%20-%20Superstore.csv"

df = pd.read_csv(url, encoding="latin1")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())
```

```python

print("Dataset shape:", df.shape)

print("\nFirst 10 rows:")
display(df.head(10))

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
print("Total transactions:", len(df))
```

```python
print(df.columns.tolist())
```

```python
missing_values = df.isnull().sum()

print("Missing values by column:")
print(missing_values[missing_values > 0])

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
duplicate_rows = df.duplicated().sum()
duplicate_ids = df["Row ID"].duplicated().sum()
unique_orders = df["Order ID"].nunique()

print("Completely duplicated rows:", duplicate_rows)
print("Duplicated Row IDs:", duplicate_ids)
print("Total transaction rows:", len(df))
print("Unique orders:", unique_orders)

```

```python
# Convert the date columns into datetime format
df["Order Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"], errors="coerce"
)

# Check the data types
print("Order Date type:", df["Order Date"].dtype)
print("Ship Date type:", df["Ship Date"].dtype)

# Check for invalid dates
print("Invalid order dates:", df["Order Date"].isna().sum())
print("Invalid ship dates:", df["Ship Date"].isna().sum())
```

```python

df["Order Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"], errors="coerce"
)

print("Order Date type:", df["Order Date"].dtype)
print("Ship Date type:", df["Ship Date"].dtype)


print("Invalid order dates:", df["Order Date"].isna().sum())
print("Invalid ship dates:", df["Ship Date"].isna().sum())
```

```python
import numpy as np


total_sales = df["Sales"].sum()


total_profit = df["Profit"].sum()


total_quantity = df["Quantity"].sum()

total_orders = df["Order ID"].nunique()

average_order_value = total_sales / total_orders

profit_margin = np.divide(
    total_profit, total_sales
) * 100

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Unique Orders: {total_orders:,}")
print(f"Average Order Value: ${average_order_value:,.2f}")
print(f"Overall Profit Margin: {profit_margin:.2f}%")
```

```python
df["Order Month"] = df["Order Date"].dt.to_period("M")

monthly_sales = (
    df.groupby("Order Month")["Sales"]
    .sum()
    .reset_index()
)


print(monthly_sales.head(12))
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(14, 6))


plt.plot(
    monthly_sales["Order Month"].astype(str),
    monthly_sales["Sales"],
    marker="o"
)


plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Order Month", fontsize=12)
plt.ylabel("Total Sales ($)", fontsize=12)


plt.xticks(rotation=90)

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python

category_analysis = (
    df.groupby("Category")[["Sales", "Profit"]]
    .sum()
    .reset_index()
)

category_analysis = category_analysis.sort_values(
    by="Sales", ascending=False
)

print(category_analysis)
```

```python
# Calculate profit margin for each category
category_analysis["Profit Margin (%)"] = (
    category_analysis["Profit"] / category_analysis["Sales"]
) * 100

# Display the results
print(category_analysis.round(2))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_analysis,
    x="Category",
    y="Profit Margin (%)"
)

plt.title("Profit Margin by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Profit Margin (%)")
plt.tight_layout()
plt.show()
```

```python

discount_analysis = (
    df.groupby("Discount")[["Sales", "Profit"]]
    .mean()
    .reset_index()
)


print(discount_analysis.round(2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(11, 6))

sns.barplot(
    data=discount_analysis,
    x="Discount",
    y="Profit"
)

plt.axhline(y=0, color="red", linestyle="--")

plt.title("Average Profit by Discount Level")
plt.xlabel("Discount Rate")
plt.ylabel("Average Profit")
plt.tight_layout()
plt.show()
```

```python

subcategory_analysis = (
    df.groupby("Sub-Category")[["Sales", "Profit"]]
    .sum()
    .reset_index()
)


subcategory_analysis = subcategory_analysis.sort_values(
    by="Profit", ascending=True
)

print(subcategory_analysis.round(2))
```

```python

region_analysis = (
    df.groupby("Region")[["Sales", "Profit"]]
    .sum()
    .reset_index()
)

region_analysis = region_analysis.sort_values(
    by="Sales", ascending=False
)

print(region_analysis.round(2))
```

```python

df["Order Year"] = df["Order Date"].dt.year


yearly_analysis = (
    df.groupby("Order Year")[["Sales", "Profit"]]
    .sum()
    .reset_index()
)

yearly_analysis["Profit Margin (%)"] = (
    yearly_analysis["Profit"] / yearly_analysis["Sales"]
) * 100

print(yearly_analysis.round(2))
```

```python

plt.figure(figsize=(10, 6))

plt.plot(
    yearly_analysis["Order Year"],
    yearly_analysis["Sales"],
    marker="o",
    label="Sales"
)

plt.plot(
    yearly_analysis["Order Year"],
    yearly_analysis["Profit"],
    marker="s",
    label="Profit"
)

plt.title("Yearly Sales and Profit Trends")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.xticks(yearly_analysis["Order Year"])
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
